# NLP in Deep Learning

### Type 1
Using `tensorflow`, `keras`, `pytorch`

```python
%load_ext cudf.pandas
import pandas as pd
import numpy as np

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, Sequential

import warnings
warnings.filterwarnings('ignore')
```

```python
vectorizer = layers.TextVectorization(max_tokens=1000, output_sequence_length=10)
x_train = ["I love this movie", "This movie was terrible", "Great acting and plot"]
y_train = np.array([1,0,1])
```

```python
vectorizer.adapt(x_train)
x_train = vectorizer(np.array(x_train))
```

```python
x_train
```

```python
model = keras.Sequential([
    keras.Input(shape=(10,)),
    layers.Embedding(input_dim=1000, output_dim=64),
    layers.LSTM(32),
    layers.Dense(1, activation='sigmoid')
])

model.compile(
    optimizer = keras.optimizers.Adam(),
    loss = keras.losses.BinaryCrossentropy(),
    metrics = ['accuracy']
)

model.fit(
    x_train, y_train,
    epochs = 5
)
```

```python
test_vec = vectorizer(np.array(["The film was awesome"]))
pred = model.predict(test_vec)
print(pred) 
```
