### **Working of Feed-forward Layer**

Let's walk through the exact calculations using a simplified numerical example. 

To keep the math easy to follow, we will use these dimensions:
* Input vector size ($d_{\text{model}}$) = **2** (instead of 512)
* Intermediate size ($d_{ff}$) = **3** (instead of 2048)


### 1. The Inputs
Let's assume our input vector $x$ (for a single word) is:
$$x = \begin{pmatrix} 2.0 & -1.0 \end{pmatrix}$$

#### First Layer Parameters:
* **Weight Matrix $W_1$** (size $2 \times 3$):
  $$W_1 = \begin{pmatrix} 1.0 & 2.0 & -1.0 \\ 0.5 & -0.5 & 3.0 \end{pmatrix}$$
* **Bias vector $b_1$** (size $3$):
  $$b_1 = \begin{pmatrix} 0.1 & -0.2 & 0.5 \end{pmatrix}$$

#### Second Layer Parameters:
* **Weight Matrix $W_2$** (size $3 \times 2$):
  $$W_2 = \begin{pmatrix} 2.0 & 0.5 \\ -1.0 & 1.0 \\ 0.0 & -2.0 \end{pmatrix}$$
* **Bias vector $b_2$** (size $2$):
  $$b_2 = \begin{pmatrix} 0.2 & -0.1 \end{pmatrix}$$


### Step A: Linear Transformation 1 (Matrix Multiplication + Bias)
First, we project $x$ from size $2$ to size $3$ using $x W_1 + b_1$:

$$\text{Product} = x W_1 = \begin{pmatrix} 2.0 & -1.0 \end{pmatrix} \begin{pmatrix} 1.0 & 2.0 & -1.0 \\ 0.5 & -0.5 & 3.0 \end{pmatrix}$$

Let's calculate each element of the product vector:
* Element 1: $(2.0 \times 1.0) + (-1.0 \times 0.5) = 2.0 - 0.5 = 1.5$
* Element 2: $(2.0 \times 2.0) + (-1.0 \times -0.5) = 4.0 + 0.5 = 4.5$
* Element 3: $(2.0 \times -1.0) + (-1.0 \times 3.0) = -2.0 - 3.0 = -5.0$

So, the product is:
$$x W_1 = \begin{pmatrix} 1.5 & 4.5 & -5.0 \end{pmatrix}$$

Now we add the bias $b_1$:
$$\text{Linear}_1 = x W_1 + b_1 = \begin{pmatrix} 1.5 & 4.5 & -5.0 \end{pmatrix} + \begin{pmatrix} 0.1 & -0.2 & 0.5 \end{pmatrix}$$
$$\text{Linear}_1 = \begin{pmatrix} 1.6 & 4.3 & -4.5 \end{pmatrix}$$


### Step B: The Activation Function (ReLU)
Next, we apply the ReLU function $\max(0, \text{Linear}_1)$ element-wise. This turns any negative numbers into `0.0`:

* $\max(0, 1.6) = 1.6$
* $\max(0, 4.3) = 4.3$
* $\max(0, -4.5) = \mathbf{0.0}$

So, our activated vector $a$ is:
$$a = \begin{pmatrix} 1.6 & 4.3 & 0.0 \end{pmatrix}$$


### Step C: Linear Transformation 2 (Matrix Multiplication + Bias)
Finally, we project the activated vector $a$ (size 3) back to size 2 using $a W_2 + b_2$:

$$\text{Product} = a W_2 = \begin{pmatrix} 1.6 & 4.3 & 0.0 \end{pmatrix} \begin{pmatrix} 2.0 & 0.5 \\ -1.0 & 1.0 \\ 0.0 & -2.0 \end{pmatrix}$$

Let's calculate each element of this final product:
* Element 1: $(1.6 \times 2.0) + (4.3 \times -1.0) + (0.0 \times 0.0) = 3.2 - 4.3 + 0 = -1.1$
* Element 2: $(1.6 \times 0.5) + (4.3 \times 1.0) + (0.0 \times -2.0) = 0.8 + 4.3 + 0 = 5.1$

So, the product is:
$$a W_2 = \begin{pmatrix} -1.1 & 5.1 \end{pmatrix}$$

Now we add the final bias $b_2$:
$$\text{Output} = a W_2 + b_2 = \begin{pmatrix} -1.1 & 5.1 \end{pmatrix} + \begin{pmatrix} 0.2 & -0.1 \end{pmatrix}$$
$$\mathbf{\text{Output} = \begin{pmatrix} -0.9 & 5.0 \end{pmatrix}}$$


### Summary
We fed the vector $\begin{pmatrix} 2.0 & -1.0 \end{pmatrix}$ into the Feed-Forward layer, and after processing it through two matrix multiplications and a non-linear activation, the output vector is: $\mathbf{\begin{pmatrix} -0.9 & 5.0 \end{pmatrix}}$
