### **Working of Embeddings before sending it to Encoders**

Let's trace how the sentence **`"I love AI"`** goes through the Transformer encoder embedding process step-by-step. 

To keep the math easy to read, we will pretend our model dimension is $d_{\text{model}} = 4$ instead of $512$.


### Step 1: Tokenization and Indexing
The sentence is split into tokens and converted into vocabulary indices.
* **Tokens**: `["I", "love", "AI"]`
* **Indices (from vocabulary)**: `[12, 105, 8]`


### Step 2: Input Embedding Lookup
The model looks up the indices in its trained **Embedding Matrix** (which has size `[vocab_size, 4]`). Let's assume the lookup returns these vectors:

* **"I" (Index 12)** $\rightarrow$ `[ 0.15, -0.42,  0.89,  0.02]`--------------------------[RANDOM NUMBERS]
* **"love" (Index 105)** $\rightarrow$ `[-0.78,  0.31, -0.11,  0.55]`----------------------[RANDOM NUMBERS]
* **"AI" (Index 8)** $\rightarrow$ `[ 0.64,  0.92, -0.34, -0.81]`--------------------------[RANDOM NUMBERS]




### Step 3: Scaling the Embeddings
We multiply each value by $\sqrt{d_{\text{model}}} = \sqrt{4} = \mathbf{2}$.

* **"I"**: `[ 0.15, -0.42,  0.89,  0.02] * 2` $\rightarrow$ **`[ 0.30, -0.84,  1.78,  0.04]`**
* **"love"**: `[-0.78,  0.31, -0.11,  0.55] * 2` $\rightarrow$ **`[-1.56,  0.62, -0.22,  1.10]`**
* **"AI"**: `[ 0.64,  0.92, -0.34, -0.81] * 2` $\rightarrow$ **`[ 1.28,  1.84, -0.68, -1.62]`**


### Step 4: Positional Encoding Generation
We compute a unique position vector for each position index ($pos = 0, 1, 2$) using the sine and cosine formulas. These values range from $-1$ to $+1$:

* **Position 0 ("I")** $\rightarrow$ **`[ 0.00,  1.00,  0.00,  1.00]`**
* **Position 1 ("love")** $\rightarrow$ **`[ 0.84,  0.54,  0.01,  0.99]`**
* **Position 2 ("AI")** $\rightarrow$ **`[ 0.91, -0.42,  0.02,  0.98]`**





### Step 5: Element-wise Summation (Final Step)
We add the **Scaled Embedding** (from Step 3) and the **Positional Encoding** (from Step 4) together:

$$\text{Final Input Vector} = \text{Scaled Embedding} + \text{Positional Encoding}$$

#### For the word "I" (pos = 0):
$$[0.30, -0.84, 1.78, 0.04] + [0.00, 1.00, 0.00, 1.00] = \mathbf{[0.30, 0.16, 1.78, 1.04]}$$

#### For the word "love" (pos = 1):
$$[-1.56, 0.62, -0.22, 1.10] + [0.84, 0.54, 0.01, 0.99] = \mathbf{[-0.72, 1.16, -0.21, 2.09]}$$

#### For the word "AI" (pos = 2):
$$[1.28, 1.84, -0.68, -1.62] + [0.91, -0.42, 0.02, 0.98] = \mathbf{[2.19, 1.42, -0.66, -0.64]}$$
