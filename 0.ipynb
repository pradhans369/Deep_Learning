```python
import numpy as np
from tensorflow import keras
from tensorflow.keras import layers
```

```python
words = ["cat", "dog", "apple", "banana"]


vectorizer = layers.TextVectorization()
vectorizer.adapt(words)

vocab = vectorizer.get_vocabulary()
embedding = layers.Embedding(input_dim=len(vocab), output_dim=6)

word_ids = vectorizer(np.array(words))
word_vectors = embedding(word_ids)

print(word_vectors.numpy())
```
