# **Positional Encoding**

**Positional encoding** is a technique used in Transformer models to inject information about the order or position of tokens into their input representations.

> The **positonal encoding** of each input word it us used before sending the input to the self-attention layer. 

Because self-attention processes all tokens in a sequence simultaneously and in parallel (it is permutation-invariant), the model has no inherent sense of word order on its own. Positional encoding solves this by assigning a unique position-dependent vector to each token's position and adding it directly to the token's semantic embedding before passing it into the attention layers.

For each input word we can use one positonal encoding like we can put `sin(postion_of_word)` to the end of the embedding. This is still not suffiect for large sentences (coz position of two different sentences might produce the same positonal number), so we generally use two positonal encoding like `sin(postion_of_word)` and `cos(postion_of_word)`.

Even if that is no sufficient we can use:
- Two Stage : `sin(postion_of_word)` and `cos(postion_of_word)`                                                                     **Recomended**
- Four Stage : `sin(postion_of_word)` and `cos(postion_of_word)` and `sin(postion_of_word/2)` and `cos(postion_of_word/2)`          **Recomended for Extremely Large sentenes**
- Six Stage : `sin(postion_of_word)` and `cos(postion_of_word)` and `sin(postion_of_word/2)` and `cos(postion_of_word/2)` and `sin(postion_of_word/3)` and `cos(postion_of_word/3)`                                                 **Recomended for Austoronomically Large Sentences**
- ans so on


**REMEMBER :** the more the positional encoding you use, more wil be the computaions and slower will be the training.

But one drawback on the above technique is would **increase the dimentionality** exponentially. So to use positional encoding without increasing the dimension of the embedding vector is, we make a new vector of **positional encoding** with the same number of dimentions as the dimentions of the embedding vector before sending input. Then we add the both the positional encoding vector and the embedding vector and produce a new vector with same number of dimentions as the embedding vector, without increasing the dimentionality.
